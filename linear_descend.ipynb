```python
import numpy as np
```

```python
X = np.array([
    [1, 1, 1],
    [1, 2, 4],
    [1, 3, 9],
    [1, 4, 16]
])

Y = np.array([
    [1],
    [2],
    [4],
    [7]
])
```

```python
np.linalg.inv(X.T.dot(X)).dot(X.T.dot(Y))
```

```python
def calc_loss(x):
    delta = X @ x - Y + alpha * x.T @ x
    return delta.T @ delta
```

```python
lr = 0.001
```

```python
alpha = 0.0001
```

```python
x0 = np.random.random(3)[:, None]
```

```python
for p in range(100000):
    step = 2 * X.T @ (X @ x0 - Y) + 2 * alpha * x0 
    x0 = x0 - step * lr
    if p % 1000 == 0:
        print(calc_loss(x0))
```
